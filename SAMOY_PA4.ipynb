```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('board2.xlsx')
df
```

#### A. VISAYAS COMMUNICATION DATAFRAME

```python
df['Average'] = df[['Math', 'GEAS', 'Electronics', 'Communication']].mean(axis=1)
```

```python
visayas_comm = df.loc[
    (df['Hometown'] == 'Visayas') &
    (df['Track'] == 'Communication')
]

VisComm = pd.DataFrame(
    visayas_comm,
    columns=['Name', 'Gender', 'Math', 'Electronics', 'Average']
)

VisComm
```

```python
print("Number of rows:", len(VisComm))
```

#### B. VISAYAS COMMUNICATION DATAFRAME

```python
visayas_female = df.loc[
    (df['Hometown'] == 'Visayas') &
    (df['Gender'] == 'Female')
]
```

```python
VisFemale = pd.DataFrame(
    visayas_female,
    columns=['Name', 'Track', 'GEAS', 'Electronics', 'Average']
)

VisFemale
```

#### C. CATEGORY-AVERAGE VISUALIZATION

```python
track_average = df.pivot_table(
    index=['Track'],
    values='Average'
).reset_index()

track_average
```

```python
gender_average = df.pivot_table(
    index=['Gender'],
    values='Average'
).reset_index()

gender_average
```

```python
hometown_average = df.pivot_table(
    index=['Hometown'],
    values='Average'
).reset_index()

hometown_average
```

```python
plt.figure(figsize=(20,5))

plt.subplot(1,3,1)
plt.bar(track_average['Track'], track_average['Average'])
plt.title('Average Grades per Track')
plt.xlabel('Track')
plt.ylabel('Mean Average Grade')

plt.subplot(1,3,2)
plt.bar(gender_average['Gender'], gender_average['Average'])
plt.title('Average Grades per Gender')
plt.xlabel('Gender')
plt.ylabel('Mean Average Grade')

plt.subplot(1,3,3)
plt.bar(hometown_average['Hometown'], hometown_average['Average'])
plt.title('Average Grades per Hometown')
plt.xlabel('Hometown')
plt.ylabel('Mean Average Grade')

plt.tight_layout()
plt.show()
```
